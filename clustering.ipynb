```python
import pandas as pd
import seaborn as sps
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('data/retailmax.csv')
```

```python
features = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)

type(scaled_features)
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=200, n_init=10, random_state=42)

clusters = kmeans.fit_predict(scaled_features)

```

```python
clusters 
len(clusters)
```

```python
df 
```

```python
df['Calif'] = clusters
df
```

```python
df['Calif'].value_counts()
```

```python
plt.figure(figsize=(10,6))
plt.figure(figsize=(10,6))
sps.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Calif', palette='Set1', s=50)
plt.title('Clusters of Customers')
plt.show()
plt.title('Clusters of Customers')
plt.show()
```
